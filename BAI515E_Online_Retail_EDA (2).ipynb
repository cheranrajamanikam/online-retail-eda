```python
import numpy as np
import pandas as pd

print("Libraries imported successfully!")
```

```python
df = pd.read_excel("Online Retail.xlsx")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
print("First five rows:")
display(df.head())

print("\nDataset dimensions:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print("Revenue column created successfully.")

display(
    df[
        [
            "InvoiceNo",
            "Description",
            "Quantity",
            "UnitPrice",
            "Revenue"
        ]
    ].head()
)
```

```python
revenue_array = df["Revenue"].dropna().to_numpy()

print("First 10 revenue values:")
print(revenue_array[:10])

print("\nFirst revenue value:")
print(revenue_array[0])

print("\nRevenue values from index 0 to 9:")
print(revenue_array[0:10])
```

```python
quantity_array = df["Quantity"].dropna().to_numpy()

print("First 10 quantities:")
print(quantity_array[:10])

print("\nFirst 10 quantities increased by 1:")
print(quantity_array[:10] + 1)

print("\nSmallest 10 quantities:")
print(np.sort(quantity_array)[:10])
```

```python
print("Total revenue:", np.sum(revenue_array))
print("Average revenue per transaction line:", np.mean(revenue_array))
print("Median revenue:", np.median(revenue_array))
print("Minimum revenue:", np.min(revenue_array))
print("Maximum revenue:", np.max(revenue_array))
print("Standard deviation:", np.std(revenue_array))
```

```python
description_series = df["Description"]

print("Product descriptions:")
display(description_series.head())
```

```python
selected_data = df[
    [
        "Description",
        "Quantity",
        "UnitPrice",
        "Revenue"
    ]
]

display(selected_data.head())
```

```python
high_revenue = df[df["Revenue"] > 500]

print("Transaction lines with revenue above £500:")
display(high_revenue.head())

print("Number of matching rows:", len(high_revenue))
```

```python
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
df["Description"] = df["Description"].fillna("Unknown Product")

df["Country"] = df["Country"].fillna("Unknown Country")

print("Missing values after handling selected columns:")
print(df[["Description", "Country"]].isnull().sum())
```

```python
df = df.dropna(
    subset=["Quantity", "UnitPrice", "InvoiceDate"]
)

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print("Missing values in essential columns:")
print(df[["Quantity", "UnitPrice", "InvoiceDate"]].isnull().sum())
```

```python
df["Description"] = df["Description"].str.strip()

display(df["Description"].head())
```

```python
df["Description"] = df["Description"].str.upper()

display(df["Description"].head())
```

```python
heart_products = df[
    df["Description"].str.contains(
        "HEART",
        na=False
    )
]

print("Products containing 'HEART':")
display(heart_products[["StockCode", "Description"]].head(10))
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["Weekday"] = df["InvoiceDate"].dt.day_name()

display(
    df[
        [
            "InvoiceDate",
            "Year",
            "Month",
            "Day",
            "Weekday"
        ]
    ].head()
)
```

```python
country_summary = df.groupby("Country").agg(
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Unique_Invoices=("InvoiceNo", "nunique"),
    Transaction_Lines=("InvoiceNo", "count")
)

country_summary = country_summary.sort_values(
    by="Total_Revenue",
    ascending=False
)

print("Country-wise retail analysis:")
display(country_summary.head(15))
```

```python
country_year_revenue = df.groupby(
    ["Country", "Year"]
)["Revenue"].sum()

print("Revenue by country and year:")
display(country_year_revenue.head(10))

first_country = country_year_revenue.index.get_level_values(
    "Country"
)[0]

print("\nYear-wise revenue for", first_country)
display(country_year_revenue.loc[first_country])
```

```python
pivot_table = pd.pivot_table(
    df,
    values="Revenue",
    index="Country",
    columns="Year",
    aggfunc="sum",
    fill_value=0
)

print("Country-wise revenue by year:")
display(pivot_table.head(15))
```

```python
high_value_transactions = df.query("Revenue > 500")

print("Transactions with revenue above £500:")
display(
    high_value_transactions[
        [
            "InvoiceNo",
            "Description",
            "Quantity",
            "UnitPrice",
            "Revenue"
        ]
    ].head(10)
)
```

```python
calculated_revenue = df.eval("Quantity * UnitPrice")

print("Revenue calculated using eval():")
display(calculated_revenue.head(10))
```

```python
print(
    "Revenue calculations match:",
    np.allclose(
        df["Revenue"].to_numpy(),
        calculated_revenue.to_numpy(),
        equal_nan=True
    )
)
```

```python
print("Descriptive statistics:")
display(
    df[
        [
            "Quantity",
            "UnitPrice",
            "Revenue"
        ]
    ].describe()
)
```

```python
top_countries = country_summary.sort_values(
    by="Total_Revenue",
    ascending=False
).head(10)

print("Top 10 countries by total revenue:")
display(top_countries)
```

```python
top_transactions = df.sort_values(
    by="Revenue",
    ascending=False
).head(10)

print("Top 10 transaction lines by revenue:")
display(
    top_transactions[
        [
            "InvoiceNo",
            "Description",
            "Quantity",
            "UnitPrice",
            "Revenue",
            "Country"
        ]
    ]
)
```
